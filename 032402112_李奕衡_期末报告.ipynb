```python
import pandas as pd
import numpy as np

# 读取数据
df = pd.read_csv('ershoufang_list.csv')

# 1. 整体市场核心指标统计
print("===== 福州二手房整体市场核心指标 =====")
core_stats = df[['面积数值', '总价数值', '均价数值']].agg([
    'count', 'mean', 'median', 'min', 'max', 'std'
]).round(2)
print(core_stats)

# 2. 行政区房源分布
print("\n===== 各行政区房源数量与占比 =====")
district_count = df['市区'].value_counts().reset_index()
district_count.columns = ['行政区', '房源数量']
district_count['市场占比'] = (district_count['房源数量'] / len(df) * 100).round(2).astype(str) + '%'
print(district_count)

# 3. 户型分布TOP10
print("\n===== 主流户型分布TOP10 =====")
house_type_top10 = df['户型'].value_counts().head(10).reset_index()
house_type_top10.columns = ['户型', '房源数量']
house_type_top10['占比'] = (house_type_top10['房源数量'] / len(df) * 100).round(2).astype(str) + '%'
print(house_type_top10)

# 4. 房龄分布
print("\n===== 房龄区间分布 =====")
age_dist = df['房龄'].value_counts().reset_index()
age_dist.columns = ['房龄区间', '房源数量']
age_dist['占比'] = (age_dist['房源数量'] / len(df) * 100).round(2).astype(str) + '%'
print(age_dist)

# 5. 各行政区价格对比
print("\n===== 各行政区均价、总价对比 =====")
district_price = df.groupby('市区').agg({
    '均价数值': ['mean', 'median'],
    '总价数值': ['mean', 'median']
}).round(2)
district_price.columns = ['均价_均值', '均价_中位数', '总价_均值', '总价_中位数']
print(district_price.sort_values('均价_中位数', ascending=False))

```

## 4. 数据质量检查

请检查缺失值、重复值、异常值、格式不统一等问题，并说明你的发现。

3. 数值异常值（面积 / 总价 / 均价）
检测发现
面积异常：极小 / 极大离群值，极小面积可能车库 / 储藏间，超大面积为别墅复式；
总价异常：部分高端改善房源总价远超四分位上限（如 300w + 鼓楼学区房），属于真实高端房源，非错误；
均价异常：鼓楼学区房源均价 3w+/㎡，为市场真实溢价，不是数据错误；
无逻辑错误（不存在面积 / 总价 / 均价≤0 的脏数据）；
处理建议
做基础统计时保留全部样本；
刚需市场专项分析可过滤极端高总价 / 大面积别墅房源。
4. 文本格式不统一 & 脏数据问题
（1）标题字段
大量标题包含多余空格、换行、特殊符号（￥、#、《》、顿号、广告乱字），如近医科大 西洋宁化地铁25年次新房中间多空格，不利于关键词提取。
（2）楼层字段
绝大多数标准中层(共31层)，少量文本存在多余空格、文字杂糅，格式不统一，无法直接提取总楼层数字。
（3）朝向字段杂乱
原始方位值：南北、南、东南、西南、东南等文本无统一标准，同一含义多种写法，分组统计会拆分样本。
（4）建造时间、户型文本
时间统一XXXX年建造格式，无明显错误；
户型全部为X室X厅X卫规范格式，无错乱文本；
（5）所属区域 / 小区

# 数据清洗与字段转换文字说明
本次针对福州二手房原始CSV数据集开展全流程清洗与结构化字段转换工作，整体分为文本脏数据清理、混合单位字段数值提取、多维度文本拆分衍生新字段、无效样本过滤四大环节：
1. **全局文本清洗**：对标题、户型、楼层、小区、区域、朝向等所有文本字段统一处理，批量清除换行符、制表符、多处连续空格、广告特殊符号（￥#&等），消除文本格式杂乱、前后冗余字符问题，统一文本干净标准，避免分组统计时因空格、符号导致同一小区/朝向被拆分为多个类别；
2. **带单位混合字段数值提取**：原始面积、总价、均价均为“数字+文字单位”混合文本，无法参与计算，通过正则表达式剥离㎡、万、元/㎡等单位，单独生成`面积数值`、`总价数值`、`均价数值`浮点型数值字段，保留原始文本字段用于溯源，新增数值字段支撑后续统计、相关性与可视化分析；
3. **复杂文本拆分，生成结构化可分析字段**：一是拆分户型文本“X室X厅X卫”，独立提取室数、厅数、卫数三个数字字段，便于按户型规模分组对比价格；二是拆解楼层文本如“中层(共31层)”，分离出楼层等级（低/中/高）、楼栋总层数两个字段，实现楼层溢价量化分析；三是标准化朝向文本，统一归类南北、东南、南向等表述，消除同类朝向多种写法造成的统计分裂；四是从建造年份文本提取4位数字`建造年份`，用于区分不同交房年份房源；
4. **脏样本过滤与重复剔除**：首先删除总价、均价存在缺失的无效房源（无价格无法开展价格分析），其次依据房源链接去重，剔除中介重复上架的同一房源，避免房源总量、小区供给统计虚高；
5. **清洗成果**：完成后数据集去除脏数据、重复与缺失价格样本，新增室数、厅数、卫数、楼层等级、总层数、标准朝向、建造年份多个结构化分析字段，全部数值变量可直接用于聚合、透视、绘图，文本字段格式统一，为后续市场量化分析提供标准化干净数据集；同时明确数据局限：仅清洗格式问题，未补充学区、装修、地铁距离等配套缺失信息，无法量化配套溢价。。

 三、数据局限与需谨慎解读的结论
 1. 数据样本局限
全部房源仅为2年内新建次新房，无老旧小区、老破小样本，本次所有结论仅适用于福州次新房市场，不能代表福州二手房整体行情，老旧学区房、老城区低价房源规律无法通过本数据推导。

 2. 价格数据局限
数据集仅为中介挂牌价，不是真实成交价格，存在业主议价空间，分析得出的均价、总价只能反映挂牌行情，无法精准代表实际成交均价、市场真实购买力，溢价、价差结论存在一定浮动空间。

 3. 配套信息缺失
数据无学区划分、地铁距离、物业、装修等级、产权年限等关键配套字段，只能通过行政区、小区间接推导学区价值，无法定量测算学区、地铁单独带来的溢价幅度，“学区是核心因素”仅为宏观定性结论，缺少量化数值支撑。

 4. 缺失房源信息影响
原始数据存在少量价格缺失样本，清洗时直接剔除；同时存在同一房源多平台重复上架的重复链接数据，去重后样本小幅缩减，房源总量、各小区供给量统计仅为近似值，不能完全等同于真实市场挂牌总量。

 5. 异常值解读提醒
鼓楼高价学区房源、大面积别墅在统计中标记为离群值，但属于真实市场房源，并非数据错误；若仅分析刚需市场，需手动剔除极端高价样本，直接使用全量数据计算均价会拉高市场整体价格中枢，刚需行情判断易产生偏差。
